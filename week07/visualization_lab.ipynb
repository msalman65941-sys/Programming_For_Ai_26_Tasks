```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt


DATA_PATH = Path("../week06/data/processed/students_clean.csv")
FIGURE_DIR = Path("figures")
FIGURE_DIR.mkdir(parents=True, exist_ok=True)


if not DATA_PATH.exists():
    DATA_PATH = Path("week06/data/processed/students_clean.csv")
    FIGURE_DIR = Path("week07/figures")
    FIGURE_DIR.mkdir(parents=True, exist_ok=True)


students = pd.read_csv(DATA_PATH)
print("Data loaded successfully!")
print(f"Dataset Shape: {students.shape} (Rows, Columns)")
```


## Part B: Trend and Relationship Charts

### B1. Line Plot — Weekly Average Attendance

**Analytical question:** How does average attendance change over the six weeks?

A line plot is suitable for showing changes over time. The weekly attendance values are plotted in chronological order to identify increases, decreases, and changes in direction.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt


weeks = [1, 2, 3, 4, 5, 6]
average_attendance = [78, 80, 81, 79, 83, 85]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(weeks, average_attendance, marker="o", color="#2563EB")

ax.set(title="Weekly Average Attendance", xlabel="Week", ylabel="Attendance (%)")

ax.set_xticks(weeks)
ax.grid(axis="y", alpha=0.3)

fig.tight_layout()
plt.show()
```


### B2. Scatter Plot — Attendance vs Final Marks

**Analytical question:** Is student attendance associated with final examination marks?

A scatter plot is used to examine the relationship between two numerical variables. Attendance is displayed on the x-axis and final marks are displayed on the y-axis.

The plot helps identify the direction and strength of the apparent relationship, possible clusters, and unusual points. Association does not prove causation.


```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.scatter(students["attendance"], students["final"], alpha=0.7, color="#0F766E")

ax.set(title="Attendance vs Final Marks", xlabel="Attendance (%)", ylabel="Final Marks")

ax.grid(alpha=0.3)

fig.tight_layout()
plt.show()
```

```python

fig, ax = plt.subplots(figsize=(7, 4))

ax.scatter(
    students["attendance"],
    students["final"],
    alpha=0.7,
    color="#0F766E"
)

ax.set(
    title="Attendance vs Final Marks",
    xlabel="Attendance (%)",
    ylabel="Final Marks"
)

ax.grid(alpha=0.3)

fig.tight_layout()
plt.show()
```


## Part C: Distributions and Comparisons

### C1. Histogram — Distribution of Weighted Scores

**Analytical question:** How are students' weighted scores distributed?

A histogram shows the frequency of numerical values within intervals called bins. Comparing different bin counts helps us understand the center, spread, skewness, clusters, and gaps in the score distribution.


```python

scores = students["weighted_score"].dropna()

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, bins in zip(axes, [5, 10, 15]):
    ax.hist(
        scores,
        bins=bins,
        color="#60A5FA",
        edgecolor="white"
    )
    ax.set(
        title=f"Weighted Scores ({bins} Bins)",
        xlabel="Weighted Score",
        ylabel="Number of Students"
    )
    ax.grid(axis="y", alpha=0.25)

fig.tight_layout()
plt.show()
```

```python

fig, ax = plt.subplots(figsize=(7, 4))

ax.hist(
    scores,
    bins=10,
    color="#60A5FA",
    edgecolor="white"
)

ax.set(
    title="Distribution of Weighted Scores",
    xlabel="Weighted Score",
    ylabel="Number of Students"
)

ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
plt.show()
```


### C2. Box Plot — Weighted Score Summary

**Analytical question:** What are the median, spread, and potential outliers of students' weighted scores?

A box plot summarizes a numerical distribution. The line inside the box represents the median, the box represents the interquartile range, and the whiskers show the range of values within the plot's outlier rule. Points beyond the whiskers may be potential outliers that require investigation.


```python

fig, ax = plt.subplots(figsize=(7, 3.5))

ax.boxplot(
    students["weighted_score"].dropna(),
    vert=False
)

ax.set(
    title="Weighted Score Summary",
    xlabel="Weighted Score"
)

ax.grid(axis="x", alpha=0.25)

fig.tight_layout()
plt.show()
```


### C3. Bar Chart — Average Weighted Score by Program

**Analytical question:** Which program has the highest observed average weighted score?

A bar chart is suitable for comparing numerical summaries across distinct categories. We calculate the mean weighted score for each program and sort the results before plotting them.


```python

program_scores = (
    students.groupby("program")["weighted_score"]
    .mean()
    .sort_values()
)

print("Average weighted score by program:")
print(program_scores)
```

```python

fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(
    program_scores.index,
    program_scores.values,
    color="#F59E0B"
)

ax.set(
    title="Average Weighted Score by Program",
    xlabel="Program",
    ylabel="Average Weighted Score"
)

ax.tick_params(axis="x", rotation=25)
ax.grid(axis="y", alpha=0.25)

fig.tight_layout()
plt.show()
```


## Part D: Build and Export a Dashboard

### D1. Student Performance Dashboard

**Objective:** Create a single dashboard containing four charts to summarize student performance.

The dashboard includes:
1. A histogram showing the distribution of weighted scores.
2. A scatter plot showing the relationship between attendance and final marks.
3. A bar chart comparing average weighted scores by program.
4. A box plot summarizing weighted scores and highlighting potential outliers.

Using a single figure makes it easier to compare different aspects of the dataset.


```python

fig, axes = plt.subplots(2, 2, figsize=(11, 8))

# 1. Histogram
axes[0, 0].hist(
    students["weighted_score"].dropna(),
    bins=10,
    color="#60A5FA",
    edgecolor="white"
)
axes[0, 0].set(
    title="Score Distribution",
    xlabel="Weighted Score",
    ylabel="Number of Students"
)

# 2. Scatter plot
axes[0, 1].scatter(
    students["attendance"],
    students["final"],
    alpha=0.7,
    color="#0F766E"
)
axes[0, 1].set(
    title="Attendance vs Final",
    xlabel="Attendance (%)",
    ylabel="Final Marks"
)

# 3. Bar chart
axes[1, 0].bar(
    program_scores.index,
    program_scores.values,
    color="#F59E0B"
)
axes[1, 0].set(
    title="Average Score by Program",
    xlabel="Program",
    ylabel="Average Weighted Score"
)
axes[1, 0].tick_params(axis="x", rotation=25)

# 4. Box plot
axes[1, 1].boxplot(
    students["weighted_score"].dropna(),
    vert=False
)
axes[1, 1].set(
    title="Weighted Score Summary",
    xlabel="Weighted Score"
)

# Gridlines
for ax in axes.flat:
    ax.grid(alpha=0.2)

fig.suptitle(
    "Student Performance Dashboard",
    fontsize=15,
    fontweight="bold"
)

fig.tight_layout(rect=[0, 0, 1, 0.96])

plt.show()
```


### D2. Save and Verify Dashboard Outputs

The dashboard is exported in PNG and PDF formats. The PNG file is saved at 300 DPI for clear presentation, while the PDF preserves vector detail for supported chart elements.

Both files are checked to confirm that they exist and are not empty.


```python

FIGURE_DIR = Path("week07/figures")
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

png_path = FIGURE_DIR / "student_dashboard.png"
pdf_path = FIGURE_DIR / "student_dashboard.pdf"

fig.savefig(png_path, dpi=300, bbox_inches="tight")
fig.savefig(pdf_path, bbox_inches="tight")

plt.close(fig)

assert png_path.exists() and png_path.stat().st_size > 0
assert pdf_path.exists() and pdf_path.stat().st_size > 0

print("PNG saved:", png_path)
print("PDF saved:", pdf_path)
print("PNG size:", png_path.stat().st_size, "bytes")
print("PDF size:", pdf_path.stat().st_size, "bytes")
print("Both files verified successfully!")
```
